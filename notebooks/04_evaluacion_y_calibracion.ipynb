## Bandas marginales
Contrastar cobertura nominal con observada. Los horizontes largos tienen pocos orígenes; el número de provincias no equivale a observaciones independientes. Infracobertura implica que las bandas deben interpretarse con especial cautela.

```python
coverage = pd.read_csv(REPORTS / 'interval_coverage.csv')
display(coverage)
ax = coverage.pivot(index='horizon', columns='nominal_coverage', values='empirical_coverage').plot(marker='o', ylim=(0, 1), figsize=(10, 4))
ax.axhline(.8, linestyle='--', color='grey')
ax.axhline(.95, linestyle='--', color='grey')
ax.set_ylabel('Cobertura empírica')
plt.show()
display(pd.read_csv(REPORTS / 'evaluation_paired_uncertainty.csv'))
```
